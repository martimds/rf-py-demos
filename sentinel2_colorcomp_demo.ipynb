# Autenticação do Earth Engine

```python
import ee
import geemap

ee.Authenticate()
ee.Initialize(project="YOUR_PROJECT_ID")
```

# Busca e visualização da imagem Sentinel-2

```python
# Região de interesse (Sinop-MT)
roi = ee.Geometry.Rectangle([-55.62, -11.95, -55.40, -11.78])

# Buscar imagem com menor cobertura de nuvens
image = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(roi)
    .filterDate("2020-01-01", "2020-12-31")
    .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 10))
    .sort("CLOUDY_PIXEL_PERCENTAGE")
    .first()
    .clip(roi)
)

print("ID:", image.get("system:index").getInfo())
```

```python
# Mapa interativo com composições coloridas
Map = geemap.Map()
Map.centerObject(roi, 12)
Map.addLayer(image, {"bands": ["B4", "B3", "B2"], "min": 0, "max": 3000}, "Cor Verdadeira")
Map.addLayer(image, {"bands": ["B8", "B4", "B3"], "min": 0, "max": 3000}, "Falsa Cor")
Map
```
